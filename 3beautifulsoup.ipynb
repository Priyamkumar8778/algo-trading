```python

import requests 
from bs4 import BeautifulSoup
import time
import yfinance as yf
```

```python
url = 'https://finance.yahoo.com/quote/DOGE-USD/'
```

```python
# Add headers to mimic a browser
headers = {
    'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36'
}

# Add delay to avoid rate limiting
time.sleep(2)  # Wait 2 seconds before request

# Make request with headers
yresponse = requests.get(url, headers=headers)
```

```python
# Check status
print(f"Status Code: {yresponse.status_code}")

# If still 429, you might need:
# - Longer delays (5-10 seconds)
# - Use a proxy service
# - Consider Yahoo's official API instead of scraping
```
