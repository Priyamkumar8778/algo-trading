```python
#request library
import requests
```

```python
url = 'https://www.amazon.in/Sandalwood-Saffron-300-Handmade-Whitening/dp/B09S6M7JQJ/ref=zg_bs_c_beauty_d_sccl_1/520-7049836-1370552?pd_rd_w=ZaOTh&content-id=amzn1.sym.b908f532-cbe7-4274-8b24-b671acc58bd2&pf_rd_p=b908f532-cbe7-4274-8b24-b671acc58bd2&pf_rd_r=603JD0HRTJ8HF1ZC835T&pd_rd_wg=CqmEt&pd_rd_r=e237d104-48c0-4378-a2d7-5508d0fa1a20&pd_rd_i=B09S6M7JQJ&th=1'
```

```python
aresponse = requests.get(url)
```

```python
aresponse
```

```python
#we are getting 500 error because amazon is blocking our request. 
# To avoid this we need to add headers to our request. 
# Headers are used to identify the browser and the operating system of the user. 
# We can use the headers to make our request look like a request from a browser.

#we have to change the user agent in the headers to make our request look like a request from a browser.
```

```python
user_agent = 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36'
```

```python
new_aresponse = requests.get(url, headers={'User-Agent': user_agent})
```

```python
new_aresponse
```

```python
#now it is working fine. 
# We are getting 200 status code which means our request is successful.
```

```python
# new_aresponse.content
for k,v in new_aresponse.headers.items():
    print(k,':',v)
```
